```python
class Kucing:
    # Constructor
    def __init__(self, nama, warna):
        self.nama = nama    # Attribute
        self.warna = warna  # Attribute

    # Method
    def suara(self):
        return f"{self.nama} berkata: Meow!"

# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih")
kucing2 = Kucing("Garfield", "Oranye")

# Mengakses Attribute dan Method
print(kucing1.nama)       # Output: Luna
print(kucing2.suara())    # Output: Garfield berkata: Meow!
```

```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name              # Nama Hero
        self.hp = hp                  # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2) # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)
eudora.info()
eudora.serang(balmond)       # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage



# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di
        # dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar
        # aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999     # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp)     # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(-50)       # Coba set negatif
print(hero1.get_hp())   # Output: 0 (Karena dicegat oleh logika Setter)

```

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.

class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):


    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
# unit = GameUnit() # ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")
# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]
print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```


## Tugas Analisis 1
Apa yang terjadi jika `hero1.hp` diubah jadi 500?

Jawaban: Kalau saya jalankan `hero1.hp = 500` lalu `print(hero1.hp)`, hasilnya **500**, bukan 100 lagi. Ini bisa terjadi karena atribut `hp` di Latihan 1 masih **publik** (belum diberi tanda `__`), jadi bisa diubah langsung dari luar class tanpa aturan apa pun.

## Tugas Analisis 2
Mengapa parameter `lawan` harus berupa objek, bukan string nama?

Jawaban: Karena kalau `lawan` hanya berupa string, saya tidak bisa memanggil `lawan.diserang(...)` di dalam method `serang()` — string tidak punya method itu. Dengan mengirim objek utuh, method bisa langsung mengakses atribut (`lawan.name`) dan memanggil method (`lawan.diserang()`) milik objek lawan. Ini bentuk nyata objek saling berinteraksi lewat method masing-masing.

## Tugas Analisis 3
**1 — Eksperimen `super()`:**
Kalau baris `super().__init__(name, hp, attack_power)` dihapus, memanggil `eudora.info()` akan error:
`AttributeError: 'Mage' object has no attribute 'name'`
Ini terjadi karena `__init__` milik `Mage` meng-override total `__init__` milik `Hero`, jadi baris `self.name = name`, `self.hp = hp`, `self.attack_power = attack_power` di `Hero` tidak pernah dijalankan. Nama `"Eudora"` masuk sebagai parameter tapi tidak pernah disimpan jadi atribut `self.name`.

**2 — Peran `super()`:**
`super()` berfungsi sebagai **jembatan** yang memanggil constructor milik class Induk dari dalam class Anak, supaya semua atribut dasar yang biasanya diisi Induk tetap terisi otomatis, tanpa perlu saya tulis ulang kodenya di class Anak.

## Tugas Analisis 4
**1 — Percobaan Hacking (`hero1._Hero__hp`):**
Nilai HP **tetap muncul**, tidak error. Ini karena Python melakukan **Name Mangling**: atribut `__hp` sebenarnya diubah namanya jadi `_Hero__hp`, bukan benar-benar disembunyikan total. Tapi ini tetap tidak boleh dilakukan dalam praktik yang baik, karena tujuan private itu supaya data hanya diubah lewat getter/setter yang sudah ada validasinya — kalau diakses paksa lewat name mangling, validasi itu jadi terlewati.

**2 — Uji Validasi (hapus if/elif di `set_hp`):**
Kalau isi `set_hp` hanya `self.__hp = nilai_baru`, lalu dipanggil `hero1.set_hp(-100)`, maka HP jadi **-100** (nilai tidak masuk akal). Ini membuktikan method Setter penting karena bukan cuma cara mengubah data, tapi juga tempat menaruh **aturan/validasi** supaya data yang masuk selalu logis.

## Tugas Analisis 5
**1 — Melanggar Kontrak (hapus method `serang` di `Hero`):**
Muncul error:
`TypeError: Can't instantiate abstract class Hero with abstract method serang`
Artinya class `Hero` dianggap belum lengkap karena belum memenuhi "janji" dari Interface `GameUnit`. Konsekuensinya, kalau method yang dijanjikan Interface tidak dibuat, class turunan itu **tidak bisa dipakai sama sekali** sampai dilengkapi.

**Poin 2 — Mencetak Cetakan (`unit = GameUnit()`):**
Ini error karena `GameUnit` adalah abstract class, sengaja dilarang dibuat objeknya langsung sebab fungsinya hanya sebagai **kerangka/kontrak**. Gunanya tetap ada meski tidak bisa jadi objek adalah untuk memaksa semua class turunan mengikuti standar method yang sama, supaya program lebih konsisten.

## Tugas Analisis 6
**1 — Uji Skalabilitas (tambah class `Healer(Hero)`):**
Program tetap **berjalan lancar** tanpa mengubah kode loop `for pahlawan in pasukan: pahlawan.serang()`. Kesimpulannya, keuntungan Polimorfisme adalah programmer bisa terus menambah karakter baru di masa depan tanpa perlu mengubah kode lama yang sudah berjalan — kode jadi lebih fleksibel dan gampang dikembangkan.

**2 — Konsistensi Penamaan (ubah `serang` jadi `tembak_panah` di `Archer`):**
Program akan error (`AttributeError`) saat loop sampai ke objek `Archer`, karena Python mencari method `serang()` yang sudah tidak ada. Nama method antar Parent dan Child harus persis sama karena Polimorfisme bekerja dengan memanggil **satu nama method yang sama** ke berbagai objek berbeda — kalau namanya beda, kode pemanggil tidak tahu harus memanggil nama apa untuk tiap objek.

```python
# Parent Class
class Hewan:
    def makan(self):
        print("Hewan ini sedang makan.")

# Child Class (Mewarisi Hewan)
class Burung(Hewan):
    def terbang(self):
        print("Burung sedang terbang.")

# Object
beo = Burung()
beo.makan()   # Mewarisi dari Hewan (Output: Hewan ini sedang makan.)
beo.terbang() # Method milik sendiri
```

```python
class AkunBank:
    def __init__(self, saldo):
        self.__saldo = saldo  # Private attribute

    def cek_saldo(self):
        return self.__saldo

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah

rekening = AkunBank(1000)
rekening.setor(500)
print(rekening.cek_saldo()) # Output: 1500
# print(rekening.__saldo) # Ini akan Error! Tidak bisa akses langsung.
```

```python
import pandas as pd
import numpy as np

# ==========================================
# TAHAP 1: Data Loading & Inspection
# ==========================================
print("--- TAHAP 1: INSPEKSI DATA ---")
# Memuat data dari file CSV
df = pd.read_csv('dataset_nilai_akademik_siswa.csv')

print(df.head()) # Menampilkan 5 baris pertama
print(df.info()) # Melihat tipe data dan data yang kosong
print(df.describe()) # Statistik ringkas
print("Ukuran dataset (baris, kolom):", df.shape)

# ==========================================
# TAHAP 2: Data Cleaning
# ==========================================
print("\n--- TAHAP 2: PEMBERSIHAN DATA ---")
# 1. Menghapus data yang duplikat
df = df.drop_duplicates()

# 2. Menangani nilai kosong (missing value)
# CATATAN: Jika nama kolom nilainya bukan 'nilai', ganti kata 'nilai' di bawah ini
df['nilai'] = df['nilai'].fillna(0)

# Jika ada kolom nama yang kosong, barisnya dihapus (hilangkan tanda # untuk menjalankan)
# df = df.dropna(subset=['nama'])

print("Pembersihan selesai! Ukuran dataset sekarang:", df.shape)
print("\n--- TAHAP 3: MANIPULASI DATA ---")


# [KODE PERBAIKAN] Menghapus kata 'poin', mengubah koma jadi titik, membersihkan spasi, lalu mengubah ke angka desimal
df['nilai'] = df['nilai'].astype(str).str.replace('poin', '', case=False).str.replace(',', '.').str.strip().astype(float)

# 1. Kolom Turunan: Membuat kolom status Lulus/Remedial (KKM = 75)
df['status'] = np.where(df['nilai'] >= 75, 'Lulus', 'Remedial')

# 2. Filtering: Menampilkan jumlah siswa yang remedial
siswa_remedial = df[df['status'] == 'Remedial']
print(f"Jumlah siswa yang remedial: {len(siswa_remedial)} anak")

# 3. Sorting: Mengurutkan 5 siswa dengan nilai tertinggi
ranking = df.sort_values(by='nilai', ascending=False)
print("\nTop 5 Siswa dengan Nilai Tertinggi:")
print(ranking[['nama', 'kelas', 'nilai']].head())

# 4. Groupby: Menghitung rata-rata nilai per kelas
rata_kelas = df.groupby('kelas')['nilai'].mean().reset_index()
print("\nRata-rata Nilai per Kelas:")
print(rata_kelas)

# ==========================================
# TAHAP 4: Menyimpan Dataset Bersih
# ==========================================
df.to_csv('dataset_nilai_bersih.csv', index=False)
print("\nDataset bersih telah berhasil disimpan menjadi 'dataset_nilai_bersih.csv'")
```
